### **Understand the data - Investigate the variables**

In this phase, the focus shifts to a deeper investigation of the variables to gain a better understanding of the dataset.

Since the ultimate goal of the project is to classify videos as either claims or opinions using machine learning, a logical first step is to examine the claim_status variable. By analyzing the distribution of this variable, insights can be gained into the balance between claims and opinions. The analysis begins by determining the number of videos for each unique claim status.

```python
# What are the different values for claim status and how many of each are in the data?
data['claim_status'].value_counts()
```


The distribution of claim statuses is fairly even.

Next, the engagement trends associated with each claim status are examined.

Boolean masking is used to filter the data based on claim status, followed by the calculation of the mean and median view counts for each claim status.

```python
# What is the average view count of videos with "claim" status?
claims = data[data['claim_status'] == 'claim']
print("mean view count for claims was", claims["video_view_count"].mean())
print("median view count for claims was", claims["video_view_count"].median())
```

```python
# What is the average view count of videos with "opinion" status?
opinions = data[data['claim_status'] == 'opinion']
print("mean view count for opinions was", opinions["video_view_count"].mean())
print("median view count for opinions was", opinions["video_view_count"].median())
```

The mean and median values for both claims and opinions are similar, indicating that the distribution of these variables is symmetrical. The view count distribution appears to be fairly balanced.

Next, trends associated with the author's ban status are examined.


```python
# Get counts for each group combination of claim status and author ban status


data.groupby(['claim_status', 'author_ban_status']).count()[['#']]
data.groupby(['claim_status', 'author_ban_status'])["video_view_count"].describe()
```

There are significantly more claim videos associated with banned authors compared to opinion videos with banned authors. This may suggest that claim videos are more closely monitored, and authors may face stricter rules when posting claims than when posting opinions. However, it is unclear whether claim videos are inherently more likely to result in bans, or if authors of claim videos are more prone to violating terms of service.

Content moderation policies may lead to higher bans for claim videos, as these videos tend to involve controversial or misinformation-prone content, which attracts more scrutiny. This increased attention could lead to a higher likelihood of a ban for authors of claim videos compared to those posting opinions, which are generally less contentious.

It is important to note that conclusions can be drawn about the status of banned versus active authors, but not about the banned videos themselves. It is impossible to determine if a specific video caused the ban, as authors may have posted both compliant and banned videos.

The investigation into engagement levels continues, now focusing on author_ban_status.

The median video share count for each author ban status is calculated.

```python
data.groupby(['author_ban_status']).agg({'video_view_count':['mean', 'median'],
                                        'video_like_count':['mean', 'median'],
                                        'video_share_count':['mean', 'median']})

```

```python
# What's the median video share count of each author ban status?
data.groupby(['author_ban_status']).agg({'video_share_count':['median']})
```

Banned authors have a median share count that is 33 times higher than that of active authors. When analyzing the views, likes, and shares, it becomes apparent that videos associated with banned or under review authors experience significantly more interactions than those of active authors. This suggests that videos from banned or under review authors may attract more attention.



```python
data.groupby(['author_ban_status']).median(numeric_only=True)[
['video_share_count']]
```

Banned authors and those under review receive significantly more views, likes, and shares compared to active authors. In most groups, the mean is notably higher than the median, indicating the presence of videos with exceptionally high engagement levels.


I created three new columns to better understand engagement rates:

- `likes_per_view`: This column represents the number of likes divided by the number of views for each video.
- `comments_per_view`: This column represents the number of comments divided by the number of views for each video.
- `shares_per_view`: This column represents the number of shares divided by the number of views for each video.

```python
# Create a likes_per_view column
data['likes_per_view'] = data['video_like_count']/ data['video_view_count']


# Create a comments_per_view column
data['comments_per_view'] = data['video_comment_count']/ data['video_view_count']

# Create a shares_per_view column
data['shares_per_view'] = data['video_share_count']/ data['video_view_count']
```

The data was organized by claim status and author ban status to analyze engagement patterns. Summary statistics, including count, mean, and median, were calculated for the newly created engagement rate columns. This provided insights into how different types of content and author statuses influence audience interactions.

```python
data.groupby(['claim_status', 'author_ban_status']).agg({'likes_per_view': ['count','mean', 'median'],
                                                       'comments_per_view' : ['count','mean', 'median'],
                                                       'shares_per_view': ['count','mean', 'median']})
```

Videos created by banned authors and those under review tend to receive significantly more views, likes, and shares than those by active authors. However, engagement rates per view appear to be influenced more by claim status than by author ban status.

Claim videos not only achieve higher view counts than opinion videos but also receive more favorable engagement, with higher average like rates. Additionally, claim videos generate more interactions through comments and shares compared to opinion videos.

For claim videos, banned authors exhibit slightly higher likes-per-view and shares-per-view rates than active authors or those under review. In contrast, for opinion videos, active authors and those under review experience higher engagement rates across all categories compared to banned authors.
